#### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Creating New Feature Columns</span>

##### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Creating Runtime (Minutes) Column</span>

The `runtime` column in the dataset contains durations in different formats, such as "1h", "30 min", and "22 min". This column is going to be converted into a consistent format showing the total runtime in minutes which will be in a new column named `runtime_minutes`.

```python
def convert_runtime_to_minutes(runtime_str):
    """ Function to convert runtime strings like "1h 30 min' or '45 min' into
    total minutes as an integer or None if the input is invalid. """
    if runtime_str == 'N/A':
        return None
    total_minutes = 0
    hours_match = re.search(r'(\d+)h', runtime_str)
    minutes_match = re.search(r'(\d+)\s*min', runtime_str)
    
    # Convert mins and add if hours are found (1) and add if mins are found, add to total minutes (2)
    if hours_match: #1
        total_minutes += int(hours_match.group(1)) * 60
    if minutes_match: #2
        total_minutes += int(minutes_match.group(1))
    return total_minutes

# Apply function to runtime column and create new runtime_minutes column (3) and verify the changes (4) and data type (5)
disney_data['runtime_minutes'] = disney_data['runtime'].apply(convert_runtime_to_minutes) #3
print(disney_data[['runtime', 'runtime_minutes']].head()) #4
print(f"runtime_minutes data type: {disney_data['runtime_minutes'].dtypes}") #5
```

##### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Creating Total Awards and Nominations Column</span>

The `awards` column in the dataset contains information like "5 wins & 13 nominations." A new column named `total_awards_nominations` is going to be created to show the total number of wins and nominations as a single numerical value for easier analysis.

```python
def calculate_total_awards(awards_str):
    """ Function to calculate the total number of awards and nominations
    from strings like "5 wins & 13 nominations" or "Nominated for 3 Primetime 
    Emmys. Returns the total count as an integer or None if the input is invalid. """
    
    # Return 0 if there is no awards (1), find all the numbers in string (2) and convert them to integers and sum them (3)
    if awards_str in ['No Awards', 'N/A']: #1
        return 0
    numbers = re.findall(r'\d+', awards_str) #2
    return sum(map(int, numbers)) #3

# Apply function to awards column and create new total_awards column (4) and verify the changes (5) and data type (6)
disney_data['total_awards_nominations'] = disney_data['awards'].apply(calculate_total_awards) #4
print(disney_data[['awards', 'total_awards_nominations']].head()) #5
print(f"total_awards_nominations data type: {disney_data['total_awards_nominations'].dtypes}") #6
```

##### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Creating Number Of Languages Column</span>

The `language` column has one or more languages listed for each movie or series, separated by commas. A new column called `number_of_languages` is going to be created to count the number of languages listed for each entry.

```python
# Creating a new column that counts the number of languages listed for each entry (1) and verify the changes (2) and data type (3)
disney_data['number_of_languages'] = disney_data['language'].apply(lambda languages: len(languages.split(',')) if languages != 'N/A' else 0) #1
print(disney_data[['language', 'number_of_languages']].head()) #2
print(f"number_of_languages data type: {disney_data['number_of_languages'].dtypes}") #3                                                          
```

##### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Creating Number Of Countries Column</span>

The `country` column has one or more countries per entry. The new column `number_of_countries` will be the total number of countries for each entry.

##### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Creating Number Of Genres Column</span>

The `genre` column has multiples genres for each movie or series, separated by commas. To better analyze the variety of genres, a new column called `number_of_genres` will be created to capture the total number of genres for each entry.

### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">📊 IMDb Ratings Analysis</span>

#### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">How do IMDb ratings differ between movies and series?</span>

```python
# Setting up the plot style (1) and theme colors (2)
sns.set(style = "whitegrid") #1
custom_colors = ['teal', 'deepskyblue', 'steelblue'] #2

# Create a box plot to compare IMDb ratings for movies and series
plt.figure(figsize = (10, 6))
sns.boxplot(data = cleaned_disney_dataset, x = 'type', y = 'imdb_rating', hue = 'type', palette = custom_colors, legend = False)

# Plot title (3) and axis labels (4) and show the plot (5)
plt.title("IMDb Ratings: Movies vs. Series", fontsize = 16, color = 'teal', fontweight = 'bold') #3
plt.xlabel("Type", fontsize = 12, color = 'teal') #4
plt.ylabel("IMDb Rating", fontsize = 12, color = 'teal')
plt.show() #5
```

##### <span style = "color: teal; ">Insights</span>
The box plot above shows how IMDb ratings vary across movies, series, and episodes on Disney Plus. 
- <span style = "font-weight: bold; color: teal;">Movies</span> have a median IMDb rating of around **6.5**, with some outliers below 4. The wide spread of ratings indicates a broader variation in quality.
- <span style = "font-weight: bold; color: teal;">Series</span> have a slightly higher median IMDb rating of around **7.0**,  with fewer low outliers compared to movies and a more even distribution of ratings.
- <span style = "font-weight: bold; color: teal;">Episodes</span> also have a median IMDb rating of around **6.5** but show a tighter distribution, indicating more consistent ratings compared to movies.

In summary, series and episodes on Disney Plus generally receive higher and more consistent ratings compared to movies. The lower ratings for some movies suggest that while many Disney movies are well-received, a few have not met audience expectations.

#### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">How do IMDb ratings vary by release year?</span>

```python
# Grouping by release year and calculating the average IMDb rating
average_ratings_by_year = cleaned_disney_dataset.groupby('released_year')['imdb_rating'].mean().reset_index()

# Plotting the trend
plt.figure(figsize = (12, 6))
sns.lineplot(data = average_ratings_by_year, x = 'released_year', y = 'imdb_rating',
             marker = 'o', linestyle = '-', color = 'navy')

# Adding vertical lines to separate the different periods
plt.axvline(x = 1950, color = 'teal', linestyle = '--', linewidth = 1)
plt.axvline(x = 1980, color = 'teal', linestyle = '--', linewidth = 1)
plt.axvline(x = 2010, color = 'teal', linestyle = '--', linewidth = 1)

# Adding text labels for each period with adjusted vertical positioning
plt.text(1932, 6, '(1930-1950)', fontsize = 10, color = 'teal', fontweight = 'bold')
plt.text(1955, 6, '(1950-1980)', fontsize = 10, color = 'teal', fontweight = 'bold')
plt.text(2012, 6, '(2010-2020)', fontsize = 10, color = 'teal', fontweight = 'bold')

# Adding a title (1) and axis labels (2), remove the gridlines (3) and show the visaul (4)
plt.title("Average IMDb Ratings by Release Year", fontsize = 16, color = 'teal', fontweight = 'bold') #1
plt.xlabel("Release Year", fontsize = 12, color = 'teal') #2
plt.ylabel("Average IMDb Rating", fontsize = 12, color = 'teal')
plt.grid(False) #3
plt.show() #4
```

##### <span style = "color: teal; ">Insights</span>
The line plot above shows how average IMDb ratings for Disney Plus movies and series have varied by release year. 
- <span style = "font-weight: bold; color: teal;">Early Years (1930-1950):</span> The average IMDb ratings during this period generally ranged between 6.5 and 7.7. Notable peaks occurred in 1937 (7.52) and 1947 (7.65), reflecting the positive reception of Disney's early animated features and storytelling techniques.
- <span style = "font-weight: bold; color: teal;">Mid-Century (1960-1980):</span> The ratings during this period displayed more variability, with averages fluctuating between 5.9 and 7.3. Lower-rated years, such as 1974 (5.93) and 1976 (6.03), may indicate shifts in audience preferences or content strategies that were less successful. The spike in 1980 is largely driven by the well-received Star Wars: Episode V - The Empire Strikes Back (8.70).
- <span style = "font-weight: bold; color: teal;">Recent Years (2010-2020):</span> The ratings show a relatively stable trend, averaging between 6.5 and 7.1. Years like 2014 (7.10) and 2019 (7.03) stand out with higher ratings, suggesting that recent content strategies, such as leveraging popular franchises and creating diverse content, have been well-received.

In summary, Disney Plus content has experienced fluctuations in IMDb ratings over the decades. Recent stable ratings indicate that current content strategies, including well-produced series and established franchises, are contributing to consistent audience approval.

#### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">Do certain months have higher IMDb ratings for released content?</span>

The bar plot above shows the average IMDb ratings for Disney Plus content based on the month of release. 
- The differences in IMDb ratings across months are relatively small, ranging between 6.29 and 6.92.
- November has the highest average rating of 6.92, while August has the lowest at 6.29. This slight variation might not indicate a strong relationship between release month and content quality.
- While months like May (6.89) and September (6.83) also show slightly higher ratings, these differences do not suggest a clear pattern or strategy.

In summary, there does not appear to be a strong relationship between the month of release and IMDb ratings. The ratings remain relatively consistent throughout the year, indicating that other factors, such as content type, genre, or awards, may have a more significant influence on ratings.

#### <span style = "color: teal; font-family: 'Comic Sans MS', cursive;">How closely do critic ratings (metascore) align with audience ratings (IMDb)?</span>

```python
# Drop rows with missing Metascore or IMDb ratings (1) and calculting the correlation coefficient (2)
cleaned_disney_dataset_filtered = cleaned_disney_dataset.dropna(subset = ['imdb_rating', 'metascore']) #1
correlation = np.corrcoef(cleaned_disney_dataset_filtered['imdb_rating'], cleaned_disney_dataset_filtered['metascore'])[0, 1] #2

# Create a scatter plot to visualize the relationship
plt.figure(figsize = (10, 6))
sns.scatterplot(data = cleaned_disney_dataset_filtered, x = 'metascore', y = 'imdb_rating', color = 'navy', alpha = 0.7)

# Add the correlation coefficient to the title (3), axis labels (4), and show the visaul (5)
plt.title(f"IMDb Ratings vs Metascore\nCorrelation: {correlation:.2f}", fontsize = 16, color = 'teal', fontweight = 'bold') #3
plt.xlabel("Metascore (Critic Rating)", fontsize = 12, color = 'teal') #4
plt.ylabel("IMDb Rating (Audience Rating)", fontsize = 12, color = 'teal') 
plt.show() #5

# Generate a correlation matrix (6) and display the corelation matrix as a heatmap (7) with the title (8) and show the vicual (9) 
correlation_matrix = cleaned_disney_dataset_filtered[['imdb_rating', 'metascore']].corr() #6
plt.figure(figsize = (8, 6))
sns.heatmap(correlation_matrix, annot = True, cmap = 'Blues', linewidths = 0.5, fmt = ".2f", annot_kws = {"size": 12}) #7
plt.title("Correlation Matrix: IMDb Ratings and Metascore", fontsize = 16, color = 'teal', fontweight = 'bold') #8
plt.show() #9
```

##### <span style = "color: teal; ">Insights</span>
The scatter plot and correlation matrix highlight the relationship between IMDb ratings (audience ratings) and Metascore (critic ratings) for Disney Plus content.
- The correlation coefficient of 0.72 indicates a strong positive correlation between Metascore and IMDb ratings. In general, content that receives higher ratings from critics also tends to be rated higher by audiences.
- The scatter plot shows a clear upward trend, reinforcing that as critic scores increase, audience ratings typically follow a similar pattern.
- For Metascores between 40 and 70, there is some variability in IMDb ratings, suggesting occasional differences in mid-tier content. However, content with high Metascores (above 80) tends to have higher IMDb ratings (above 7), while content with low Metascores (below 40) typically receives lower IMDb ratings (below 5).

In summary, critic ratings (Metascore) and audience ratings (IMDb) are closely aligned, suggesting that critically acclaimed Disney Plus content tends to resonate well with audiences. The correlation of 0.72 reflects a consistent agreement between critics and viewers, with only slight deviations in the mid-range scores.
